```python
import os
import shutil
from glob import glob
```

```python
data_dir = "/home/sonymd/Downloads/ChestXray14Data/images"
data_sub_dir = "/home/sonymd/Downloads/ChestXray14Data/small_subset"

imgs = os.listdir(data_dir)
IDs = list(set([img.split("_")[0] for img in imgs]))

train_ids, valid_ids, test_ids = IDs[:1000], IDs[1000:1200], IDs[1200:1400]
```

```python
len(train_ids),len(valid_ids), len(test_ids)
```

```python
for img_file in imgs:
    ID = img_file.split("_")[0]
    if ID in train_ids:
        shutil.copy(os.path.join(data_dir, img_file), os.path.join(data_sub_dir, "train", img_file))
    elif ID in valid_ids:
        shutil.copy(os.path.join(data_dir, img_file), os.path.join(data_sub_dir, "valid", img_file))
    elif ID in test_ids:
        shutil.copy(os.path.join(data_dir, img_file), os.path.join(data_sub_dir, "test", img_file))

```
