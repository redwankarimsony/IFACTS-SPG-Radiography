### Finding the models and their accruacies.  

```python
import matplotlib.pyplot as plt
import os
import numpy as np
import pandas as pd
import seaborn as sns
from summarize import summarize_single_experiment

# Basic plot setup
sns.set(style="whitegrid")
sns.set_context("paper")
sns.set(font_scale=1.5)
sns.set_style("ticks")

# Read the data

EXPERIMENT_NAME = "experiments/base_experiment/"

df = summarize_single_experiment(experiment_name=EXPERIMENT_NAME, verbose=False)
print(df.head())


# Plot the of all boxes grouped by the model architecture
plt.figure(figsize=(15, 10))
sns.barplot(x='model_name', y='val_acc', hue='box_preset', data=df)
plt.xticks(rotation=70)
plt.title('Accuracy of each model grouped by Category')
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(EXPERIMENT_NAME, "summary", 'accuracy-grouped-by-models.png'))
plt.show()

# Plot the of all boxes grouped by the model architecture
plt.figure(figsize=(15, 10))
sns.barplot(x='box_preset', y='val_acc', hue='model_name', data=df)
plt.xticks(rotation=70)
plt.title('Accuracy of each model grouped by Category')
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(EXPERIMENT_NAME, "summary", 'accuracy-grouped-by-box-preset.png'))
plt.show()


```
