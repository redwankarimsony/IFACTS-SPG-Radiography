```python
import matplotlib
matplotlib.use("Agg")

from glob import glob
import matplotlib.pyplot as plt
import cv2

```

```python
corrects = glob("saved_checkpoints2/box_0/resnet34/correctly_classified_samples/*.png")
incorrects = glob("saved_checkpoints2/box_0/resnet34/misclassified_samples/*.png")

idx =0
```

```python
fig = plt.figure(figsize=(15, 15), dpi=100)
row = 4
col= 4

for i in range(1, row*col+1):
    filename = corrects[idx].split(" ")[0].split("/")[-1]
    fig.add_subplot(row, col, i)
    img = cv2.imread(corrects[idx], 1)
    plt.imshow(img, cmap=plt.cm.bone)
    plt.axis('off')
    plt.title(filename)
    idx+=1

plt.suptitle("Correctly Classified Samples")
plt.savefig("result_correct.png")
```

```python
fig = plt.figure(figsize=(15, 15), dpi=100)
row = 4
col= 4

for i in range(1, row*col+1):
    filename = incorrects[i].split(" ")[0].split("/")[-1]
    fig.add_subplot(row, col, i)
    img = cv2.imread(incorrects[i], 1)
    plt.imshow(img, cmap=plt.cm.bone)
    plt.axis('off')
    plt.title(filename + " as " + incorrects[i].split(" ")[-1].split(".")[0])

plt.suptitle("Incorrectly Classified Samples")
plt.savefig("result_incorrect.png")
```
